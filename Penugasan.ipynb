#1.	Eksperimen dengan Modul math

```python
import math
def luas_lingkaran(jari_jari):
  return math.pi*(jari_jari**2)

jari_1 = 7
luas1 = luas_lingkaran(jari_1)
print(f"Luas lingkaran dengan jari-jari {jari_1} adalah {luas1:.2f}")

jari_2 = 10
luas2 = luas_lingkaran(jari_2)
print(f"Luas lingkaran dengan jari-jari {jari_2} adalah {luas2:.2f}")
```

#4.	Modul Buatan Sendiri

```python
import konversi

def celcius_ke_fahrenheit(c):
  f = (c * 9/5) + 32
  return f

def fahrenheit_ke_celcius(f):
  c = (f - 32) * 5/9
  return c

celcius = 50
fahrenheit = celcius_ke_fahrenheit(celcius)
print(f"{celcius} Celcius sama dengan {fahrenheit:.2f} Fahrenheit")

fahrenheit = 20
celsius = fahrenheit_ke_celcius(fahrenheit)
print(f"{fahrenheit} Fahrenheit sama dengan {celsius:.2f} Celcius")
```
